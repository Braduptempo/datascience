## Business understanding
**Aluminia**, een fabrikant van aluminiumcomponenten, laat elke batch visueel beoordelen door een kwaliteitsinspecteur. Op basis van deze inspectie wordt de batch goedgekeurd of afgekeurd. Deze kwaliteitsinspecteur gaat binnenkort met pensioen. Daarom wil Aluminia onderzoeken of dit proces geautomatiseerd kan worden. Hiervoor zijn verschillende proces- en kwaliteitsindicatoren gemeten en vastgelegd in `productkwaliteit.csv`:
- proceskwaliteit: score (0–100) die de stabiliteit van het productieproces weergeeft
- grondstofscore: beoordeling van de binnenkomende grondstoffen
- temperatuurvariatie: spreiding in temperatuur tijdens de productie
- machine_trilling: gemeten trillingen (gestandaardiseerd)
- operator_id: ID van de bedienende operator

En ook nog scores met betrekking tot de bedienende operator (`operator.csv`):
- operator_id: ID van de bedienende operator
- operatorscore: score van de bedienende operator op die dag
